```python
df["income"] = df["income"].fillna("unknown values") #for filling with constant
```

```python
df["income"] = df["income"].fillna()(df["income"].mean())   #numerical and data should not contain any outliers
```

```python
df["income"] = df["income"].fillna()(df["income"].median())   #numerical and data should contain outliers
```

```python
df["income"] = df["income"].fillna()(df["income"].mode())   #numerical and text kind of data
```

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace = True)
```

```python
df.drop_duplicates(
    subset = ["income"],
    inplace = True
)
```

```python
df["Gender"] = df["Gender"].replace({
    "M":"Male",
    "F":"Female"

})
```
